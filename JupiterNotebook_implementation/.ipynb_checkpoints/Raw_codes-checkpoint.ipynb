# Chromadb intro

https://docs.trychroma.com/docs/overview/getting-started

-- Chromadb do the sementic search for the query in the db. Tie queries to the most related ids in db.

```python
pip install chromadb==0.5.0 -q # 1.0.13
```

```python
import chromadb
chroma_client = chromadb.Client()
```

```python
collection = chroma_client.get_or_create_collection("collection_1")
```

```python
collection.add(
    ids=["id1","id2"],
    documents =[
        "this the windows laptop with ssd-512",
        "this is the mack with air m1 chip"
    ]
)
```

```python
result_query = collection.query(
#     query_texts =["Hp","design and style"],
     query_texts =["Hp"],

    n_results = 2
)

# print(result_query["documents"][0])
print(result_query) 

#returns the id1 first then id2. coz id1 is more related to the given query with in the distnace of 1.3 then id2 (euclidian dist)
```

```python
import getpass
import os

if not os.environ.get("GROQ_API_KEY"):
  os.environ["GROQ_API_KEY"] = getpass.getpass("Enter API key for Groq: ")

from langchain.chat_models import init_chat_model

model = init_chat_model("llama3-8b-8192", model_provider="groq")
```

```python
response = model.invoke("hello, hi who are you?")
```

```python
print(response)
```

```python
print(response.content)
```

https://python.langchain.com/docs/integrations/chat/groq/

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="deepseek-r1-distill-llama-70b",
    temperature=0,
    max_tokens=None,
    reasoning_format="parsed",
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "I love you."),
]
ai_msg = llm.invoke(messages)
ai_msg
```

```python
ai_msg.content
```

### LLM as a language Translater

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "give answer for this query {input_language} to {output_language}",
        ),
        ("human", "{input}"),
    ]
)

chain = prompt | model
response = chain.invoke(
    {
        "input_language": str(input('from:')),
        "output_language": str(input('to:')),
        "input": str(input('sentance: ')),
    }
)

print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(

[
   ("system","anser for this queries {input}"),
    ("human","{input}")   
]
)

chain = prompt | llm        #llm (deepseek r1) ---> model (llama-3)
response = chain.invoke(
    
    {
        "input": str(input("enter query:"))
    }
)

print(response.content)
```
